```python
import os
import subprocess
from pathlib import Path
```

```python

def extract_first_frames(source_dir=".", output_dir="./first_frames"):
    """Extract first frame from each MP4 file recursively"""
    source_path = Path(source_dir)
    output_path = Path(output_dir)
    
    # Create output directory
    output_path.mkdir(exist_ok=True)
    
    # Find all MP4 files recursively
    mp4_files = source_path.rglob("*.MP4")
    
    for mp4_file in mp4_files:
        # Maintain directory structure in output
        rel_path = mp4_file.relative_to(source_path)
        output_subdir = output_path / rel_path.parent
        output_subdir.mkdir(parents=True, exist_ok=True)
        
        # Create output filename
        output_file = output_subdir / f"{mp4_file.stem}_frame1.jpg"
        
        # Extract first frame using ffmpeg
        cmd = [
            "ffmpeg", "-i", str(mp4_file),
            "-vf", "select=eq(n\\,0)",
            "-q:v", "3",
            "-frames:v", "1",
            str(output_file), "-y"
        ]
        
        try:
            subprocess.run(cmd, capture_output=True, check=True)
            print(f"✓ Extracted: {output_file}")
        except subprocess.CalledProcessError:
            print(f"✗ Failed: {mp4_file}")

if __name__ == "__main__":
    import sys
    source = sys.argv[1] if len(sys.argv) > 1 else "."
    output = sys.argv[2] if len(sys.argv) > 2 else "./first_frames"
    extract_first_frames(source, output)
```
